```python
import pandas as pd
import sqlite3
```

```python
# Load the CSV files
train_df = pd.read_csv("train.csv")
features_df = pd.read_csv("features.csv")
stores_df = pd.read_csv("stores.csv")

# Create a connection to a new SQLite database file
conn = sqlite3.connect('walmart.db')

# Load the data from the pandas DataFrames into SQL tables
print("Loading data into SQL tables...")
train_df.to_sql('sales', conn, if_exists='replace', index=False)
features_df.to_sql('features', conn, if_exists='replace', index=False)
stores_df.to_sql('stores', conn, if_exists='replace', index=False)
print("Tables 'sales', 'features', and 'stores' created successfully in walmart.db")

# Close the connection
conn.close()
```

```python
# --- Connect to the database and run the SQL query ---
conn = sqlite3.connect('walmart.db')
query = """
    SELECT
        s.Date,
        s.Store,
        s.Dept,
        s.Weekly_Sales,
        st.Type,
        st.Size,
        f.Temperature,
        f.Fuel_Price,
        f.CPI,
        f.Unemployment,
        f.IsHoliday
    FROM
        sales AS s
    JOIN
        stores AS st ON s.Store = st.Store
    JOIN
        features AS f ON s.Store = f.Store AND s.Date = f.Date;
"""

print("Executing SQL query to join and fetch data...")
df = pd.read_sql_query(query, conn)
print("Query complete. Data loaded into DataFrame.")

# --- Close the connection ---
conn.close()

# --- Display the results ---
print("\n--- Head of Final Merged DataFrame ---")
df.head()
```
